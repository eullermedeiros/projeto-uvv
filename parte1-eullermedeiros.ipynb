# Workflow: Detalhamento das Atividades

##Atividade A1: Estratificação amostral da população

###Atividade A1.1: Determinar o TAMANHO IDEAL da amostra

Criar os (4) gráficos do ERRO-PADRÃO-MÉDIO (TARGET: $y$: **TOTAL (R$)**) das seguintes categorias:

*   **ESSENTIAL**
*   **VIP**
*   **PRIME**
*   **GERAL**

Fazer a **ANÁLISE BOOTSTRAP** para determinar:
*   **Intervalo de confiança**.
*   **Erro padrão**.
*   **RESPONDER: Qual é o "Tamanho Ideal" da amostra ?**

Determinação do Tamanho Ideal da Amostra em Populações Infinitas Estratificadas: [Link](https://drive.google.com/file/d/1aLWDSNRrjr03ItMSE_3joXYmdoC09jQm/view?usp=sharing)

Usar a Função - Determinação do Tamanho Ideal da Amostra em Populações Infinitas Estratificadas, por dois(2) procedimentos a saber:

*   Procedimento empírico: Estabilização marginal do erro-padrão: **optimal_sample_size**
*   Procedimento teórico-empírico: Estabilização marginal do erro-padrão estratificado e otimização de Neyman: **optimal_stratified_sample_size**



```python
def Atividade1_1():
    """

    Atividade A1.1 — Determinação do tamanho ideal da amostra.

    OBJETIVO
    Avaliar o tamanho de amostra recomendado para estimar a média
    de TOTAL (R$), utilizando os procedimentos de estabilização
    do erro-padrão e a amostragem estratificada por categoria.

    DADOS UTILIZADOS
    Base: table18, carregada previamente do arquivo table18.csv.
    Variável quantitativa: TOTAL (R$).
    Variável de estratificação: TIPO DA CONTA.
    Categorias: ESSENTIAL, VIP e PRIME.
    A análise GERAL considera todas as categorias juntas.

    PROCEDIMENTOS

    1. Estabilização marginal do erro-padrão — optimal_sample_size
       Aplica o procedimento separadamente a ESSENTIAL, VIP, PRIME
       e GERAL, gerando um gráfico para cada análise.

       Avalia tamanhos de 100 a 150.000 registros, em incrementos
       de 100. A estabilização é confirmada após 10 reduções
       consecutivas do erro-padrão menores ou iguais a R$ 0,001.

    2. Estabilização estratificada — optimal_stratified_sample_size
       Estima a proporção e o desvio-padrão de cada categoria.
       Utiliza a alocação de Neyman para distribuir a amostra
       conforme o peso e a variabilidade de cada estrato.

       Avalia tamanhos totais de 100 a 100.000 registros, em
       incrementos de 100. A estabilização é confirmada após
       3 reduções relativas consecutivas do erro-padrão
       inferiores a 0,001, equivalente a 0,1%.

    3. Seleção da amostra
       Sorteia registros sem reposição dentro de cada categoria,
       respeitando as quantidades calculadas pela alocação de
       Neyman. Utiliza sementes fixas para permitir reprodução.

    4. Bootstrap estratificado
       Realiza 5.000 reamostragens com reposição dentro de cada
       categoria da amostra selecionada.

       Em cada repetição, calcula a média estratificada usando
       as proporções das categorias na base original.
       Estima o erro-padrão pelo desvio-padrão das médias bootstrap
       e o intervalo de confiança de 90% pelo método percentil.

    RESULTADOS EXIBIDOS
    - Quatro gráficos do procedimento de estabilização simples.
    - Tamanhos recomendados e respectivos erros-padrão.
    - Gráfico do procedimento estratificado.
    - Tabela de alocação da amostra entre as categorias.
    - Total de registros selecionados e contagem por categoria.
    - Histograma das médias bootstrap.
    - Média estimada, erro-padrão bootstrap e intervalo de confiança.

    RESULTADOS FINAIS — ATIVIDADE A1.1

    Intervalo de confiança de 90% para a média de TOTAL (R$): R$ 830.5481 a R$ 837.8111

    Erro-padrão bootstrap: R$ 2.1977

    Qual é o tamanho ideal da amostra?
    - O tamanho recomendado é de 50300 registros, pelo procedimento estratificado
      com alocação de Neyman e os parâmetros de estabilização adotados.


    """

    def optimal_sample_size(
        df,
        value_column,
        n_min=100,
        n_max=50000,
        step=100,
        stabilization_threshold=0.01,
        consecutive_points=3,
        plot=True
    ):

        if not isinstance(df, pd.DataFrame):
            raise TypeError("df deve ser um DataFrame do pandas.")

        if value_column not in df.columns:
            raise KeyError(f"A coluna '{value_column}' não foi encontrada.")

        if not isinstance(n_min, int) or isinstance(n_min, bool) or n_min < 2:
            raise ValueError("n_min deve ser um inteiro maior ou igual a 2.")

        if not isinstance(n_max, int) or isinstance(n_max, bool) or n_max <= n_min:
            raise ValueError("n_max deve ser um inteiro maior que n_min.")

        if not isinstance(step, int) or isinstance(step, bool) or step <= 0:
            raise ValueError("step deve ser um inteiro positivo.")

        if (
            not isinstance(stabilization_threshold, (int, float))
            or isinstance(stabilization_threshold, bool)
            or not np.isfinite(stabilization_threshold)
            or stabilization_threshold <= 0
        ):
            raise ValueError(
                "stabilization_threshold deve ser finito e positivo."
            )

        if (
            not isinstance(consecutive_points, int)
            or isinstance(consecutive_points, bool)
            or consecutive_points < 1
        ):
            raise ValueError(
                "consecutive_points deve ser um inteiro positivo."
            )

        values = pd.to_numeric(df[value_column], errors="coerce").dropna()

        if len(values) < 2:
            raise ValueError(
                "São necessárias pelo menos duas observações numéricas válidas."
            )

        standard_deviation = float(values.std(ddof=1))

        if not np.isfinite(standard_deviation):
            raise ValueError("Não foi possível estimar o desvio-padrão.")

        if standard_deviation == 0:
            raise ValueError("A variável selecionada tem variabilidade zero.")

        sample_sizes = np.arange(n_min, n_max + 1, step, dtype=int)

        if len(sample_sizes) <= consecutive_points:
            raise ValueError(
                "O intervalo de busca deve conter mais tamanhos candidatos."
            )

        standard_errors = standard_deviation / np.sqrt(sample_sizes)
        mean_standard_error = float(np.mean(standard_errors))

        marginal_reductions = np.full(len(standard_errors), np.nan)
        marginal_reductions[1:] = (
            standard_errors[:-1] - standard_errors[1:]
        )

        stable_points = np.zeros(len(sample_sizes), dtype=bool)
        stable_points[1:] = (
            marginal_reductions[1:] <= stabilization_threshold
        )

        confirmation_index = None
        consecutive_count = 0

        for index in range(1, len(sample_sizes)):
            if stable_points[index]:
                consecutive_count += 1
            else:
                consecutive_count = 0

            if consecutive_count >= consecutive_points:
                confirmation_index = index
                break

        if confirmation_index is not None:
            start_index = confirmation_index - consecutive_points + 1

            stabilization_indices = np.arange(
                start_index, confirmation_index + 1
            )

            stabilization_start = int(sample_sizes[start_index])
            optimal_n = int(sample_sizes[confirmation_index])
            optimal_se = float(standard_errors[confirmation_index])
            optimal_reduction = float(
                marginal_reductions[confirmation_index]
            )
            stabilization_found = True

        else:
            start_index = None
            stabilization_indices = np.array([], dtype=int)
            stabilization_start = None
            optimal_n = None
            optimal_se = None
            optimal_reduction = None
            stabilization_found = False

        results = pd.DataFrame({
            "sample_size": sample_sizes,
            "standard_error": standard_errors,
            "marginal_reduction": marginal_reductions,
            "stable_point": stable_points,
            "consecutive_confirmation": False
        })

        if stabilization_found:
            results.loc[
                stabilization_indices, "consecutive_confirmation"
            ] = True

        if plot:
            figure, axis = plt.subplots(
                figsize=(15, 8),
                layout="constrained"
            )

            axis.plot(
                sample_sizes, standard_errors,
                color="#2878B5", linewidth=2.6
            )

            axis.axhline(
                y=mean_standard_error,
                color="#D62728", linestyle="--", linewidth=2
            )

            axis.text(
                n_max, mean_standard_error,
                f"  Erro-padrão médio = {mean_standard_error:.4f}",
                color="#D62728", fontsize=10,
                va="bottom", ha="right"
            )

            if stabilization_found:
                zoom_x = sample_sizes[stabilization_indices]
                zoom_y = standard_errors[stabilization_indices]

                axis.scatter(
                    zoom_x, zoom_y,
                    s=95, color="#FF9F1C",
                    edgecolor="white", linewidth=1.2, zorder=5
                )

                axis.scatter(
                    optimal_n, optimal_se,
                    s=190, color="#D62728",
                    edgecolor="white", linewidth=1.8, zorder=6
                )

                axis.axvline(
                    x=optimal_n,
                    color="#D62728", linestyle=":", linewidth=1.8
                )

                annotation = (
                    f"n recomendado = {optimal_n:,}\n"
                    f"EP = {optimal_se:.4f}\n"
                    f"ΔEP = {optimal_reduction:.4f}"
                ).replace(",", ".")

                axis.annotate(
                    annotation,
                    xy=(optimal_n, optimal_se),
                    xytext=(20, 38),
                    textcoords="offset points",
                    fontsize=10.5,
                    bbox={
                        "boxstyle": "round,pad=0.45",
                        "facecolor": "white",
                        "edgecolor": "#D62728",
                        "alpha": 0.95
                    },
                    arrowprops={
                        "arrowstyle": "->",
                        "color": "#D62728",
                        "linewidth": 1.4
                    }
                )

                inset_axis = inset_axes(
                    axis,
                    width="43%",
                    height="43%",
                    loc="upper right",
                    borderpad=1.8
                )

                context_start = max(start_index - 1, 0)
                context_end = min(
                    confirmation_index + 1, len(sample_sizes) - 1
                )
                context_indices = np.arange(
                    context_start, context_end + 1
                )

                inset_axis.plot(
                    sample_sizes[context_indices],
                    standard_errors[context_indices],
                    color="#2878B5", linewidth=2.2,
                    marker="o", markersize=5
                )

                inset_axis.scatter(
                    zoom_x, zoom_y,
                    s=90, color="#FF9F1C",
                    edgecolor="white", linewidth=1.1, zorder=5
                )

                inset_axis.scatter(
                    optimal_n, optimal_se,
                    s=145, color="#D62728",
                    edgecolor="white", linewidth=1.5, zorder=6
                )

                inset_axis.axhline(
                    y=mean_standard_error,
                    color="#D62728", linestyle="--",
                    linewidth=1.2, alpha=0.8
                )

                for point_number, point_index in enumerate(
                    stabilization_indices, start=1
                ):
                    inset_axis.annotate(
                        str(point_number),
                        xy=(
                            sample_sizes[point_index],
                            standard_errors[point_index]
                        ),
                        xytext=(0, 10),
                        textcoords="offset points",
                        ha="center", fontsize=9,
                        fontweight="bold", color="#8C4A00"
                    )

                x_margin = max(step * 0.4, 1)

                inset_axis.set_xlim(
                    sample_sizes[context_start] - x_margin,
                    sample_sizes[context_end] + x_margin
                )

                local_standard_errors = standard_errors[context_indices]
                y_range = float(
                    local_standard_errors.max()
                    - local_standard_errors.min()
                )

                if y_range == 0:
                    y_margin = max(
                        abs(float(local_standard_errors.mean())) * 0.05,
                        1e-6
                    )
                else:
                    y_margin = y_range * 0.30

                inset_axis.set_ylim(
                    float(local_standard_errors.min()) - y_margin,
                    float(local_standard_errors.max()) + y_margin
                )

                inset_axis.set_title(
                    f"Janela ampliada: {consecutive_points} pontos consecutivos",
                    fontsize=10, fontweight="bold"
                )
                inset_axis.set_xlabel("Tamanho da amostra", fontsize=8)
                inset_axis.set_ylabel("Erro-padrão", fontsize=8)
                inset_axis.tick_params(axis="both", labelsize=8)
                inset_axis.grid(
                    True, linestyle="--", linewidth=0.6, alpha=0.5
                )

                mark_inset(
                    axis, inset_axis,
                    loc1=2, loc2=4,
                    fc="none", ec="#E76F8A", linewidth=1.5
                )

                title = (
                    "Estabilização Marginal Consecutiva do Erro-Padrão\n"
                    f"Estabilização confirmada em n = {optimal_n:,}"
                ).replace(",", ".")

            else:
                title = (
                    "Estabilização Marginal Consecutiva do Erro-Padrão\n"
                    "Estabilização não encontrada no intervalo analisado"
                )

            axis.set_title(
                title, fontsize=15, fontweight="bold", pad=12
            )
            axis.set_xlabel("Tamanho total da amostra (n)", fontsize=12)
            axis.set_ylabel("Erro-padrão", fontsize=12)
            axis.grid(
                True, linestyle="--", linewidth=0.7, alpha=0.45
            )

            plt.show()

        return {
            "optimal_sample_size": optimal_n,
            "stabilization_start": stabilization_start,
            "standard_error": optimal_se,
            "mean_standard_error": mean_standard_error,
            "marginal_reduction": optimal_reduction,
            "population_standard_deviation": standard_deviation,
            "stabilization_threshold": stabilization_threshold,
            "consecutive_points": consecutive_points,
            "step": step,
            "stabilization_found": stabilization_found,
            "results": results
        }

    # Somente para ESSENTIAL.
    essential = table18[
        table18["TIPO DA CONTA"] == "ESSENTIAL"
    ].copy()

    resultado_essential = optimal_sample_size(
        df=essential,
        value_column="TOTAL (R$)",
        n_min=100,
        n_max=150000,
        step=100,
        stabilization_threshold=0.001,
        consecutive_points=10,
        plot=True
    )

    print("Categoria: ESSENTIAL")
    print("Tamanho recomendado:", resultado_essential["optimal_sample_size"])
    print("Início da estabilização:", resultado_essential["stabilization_start"])

    if resultado_essential["stabilization_found"]:
        print(f"Erro-padrão: R$ {resultado_essential['standard_error']:.4f}")
        print(f"Erro-padrão médio: R$ {resultado_essential['mean_standard_error']:.4f}")
        print(f"Redução marginal final: R$ {resultado_essential['marginal_reduction']:.4f}")


            # Somente os registros VIP.
    vip = table18[
        table18["TIPO DA CONTA"] == "VIP"
    ].copy()

    resultado_vip = optimal_sample_size(
        df=vip,
        value_column="TOTAL (R$)",
        n_min=100,
        n_max=150000,
        step=100,
        stabilization_threshold=0.001,
        consecutive_points=10,
        plot=True
    )

    print("Categoria: VIP")
    print("Tamanho recomendado:", resultado_vip["optimal_sample_size"])
    print("Início da estabilização:", resultado_vip["stabilization_start"])

    if resultado_vip["stabilization_found"]:
        print(f"Erro-padrão: R$ {resultado_vip['standard_error']:.4f}")
        print(f"Erro-padrão médio: R$ {resultado_vip['mean_standard_error']:.4f}")
        print(f"Redução marginal final: R$ {resultado_vip['marginal_reduction']:.4f}")

            # Somente os registros PRIME.
    prime = table18[
        table18["TIPO DA CONTA"] == "PRIME"
    ].copy()

    resultado_prime = optimal_sample_size(
        df=prime,
        value_column="TOTAL (R$)",
        n_min=100,
        n_max=150000,
        step=100,
        stabilization_threshold=0.001,
        consecutive_points=10,
        plot=True
    )

    print("Categoria: PRIME")
    print("Tamanho recomendado:", resultado_prime["optimal_sample_size"])
    print("Início da estabilização:", resultado_prime["stabilization_start"])

    if resultado_prime["stabilization_found"]:
        print(f"Erro-padrão: R$ {resultado_prime['standard_error']:.4f}")
        print(f"Erro-padrão médio: R$ {resultado_prime['mean_standard_error']:.4f}")
        print(f"Redução marginal final: R$ {resultado_prime['marginal_reduction']:.4f}")

            # Analisa todas as categorias juntas.
    resultado_geral = optimal_sample_size(
        df=table18,
        value_column="TOTAL (R$)",
        n_min=100,
        n_max=150000,
        step=100,
        stabilization_threshold=0.001,
        consecutive_points=10,
        plot=True
    )

    print("Categoria: GERAL")
    print("Tamanho recomendado:", resultado_geral["optimal_sample_size"])
    print("Início da estabilização:", resultado_geral["stabilization_start"])

    if resultado_geral["stabilization_found"]:
        print(f"Erro-padrão: R$ {resultado_geral['standard_error']:.4f}")
        print(f"Erro-padrão médio: R$ {resultado_geral['mean_standard_error']:.4f}")
        print(f"Redução marginal final: R$ {resultado_geral['marginal_reduction']:.4f}")


    def optimal_stratified_sample_size(
        df,
        value_column,
        strata_column,
        n_min=100,
        n_max=50000,
        step=100,
        stabilization_threshold=0.01,
        consecutive_points=3,
        plot=True
    ):

        data = df[[value_column, strata_column]].dropna().copy()

        data[value_column] = pd.to_numeric(
            data[value_column], errors="coerce"
        )
        data = data.dropna()

        if data.empty:
            raise ValueError("Não há dados válidos para analisar.")


        weights = (
            data[strata_column]
            .value_counts(normalize=True)
            .sort_index()
        )

        # Desvio-padrão de TOTAL (R$) em cada categoria.
        stds = (
            data.groupby(strata_column)[value_column]
            .std()
            .sort_index()
        )

        stds = stds.loc[weights.index]

        strata = weights.index.values
        Wh = weights.values
        Sh = stds.values

        if np.any(np.isnan(Sh)):
            raise ValueError(
                "Algum estrato não tem observações suficientes "
                "para calcular o desvio-padrão."
            )

        if np.any(Sh == 0):
            print("Aviso: existe um estrato com desvio-padrão zero.")


        neyman_weights = Wh * Sh

        if neyman_weights.sum() == 0:
            raise ValueError("Os pesos de Neyman são todos zero.")

        sample_sizes = np.arange(n_min, n_max + step, step)

        standard_errors = []
        allocations = []

        for n in sample_sizes:

            nh = n * neyman_weights / neyman_weights.sum()


            se = np.sqrt(
                np.sum((Wh ** 2) * (Sh ** 2) / nh)
            )

            standard_errors.append(se)
            allocations.append(nh)

        standard_errors = np.array(standard_errors)
        allocations = np.array(allocations)


        marginal_reduction = (
            np.abs(np.diff(standard_errors))
            / standard_errors[:-1]
        )

        stable_sequence_index = None

        for i in range(
            len(marginal_reduction) - consecutive_points + 1
        ):
            window = marginal_reduction[i:i + consecutive_points]

            if np.all(window < stabilization_threshold):
                stable_sequence_index = i
                break

        if stable_sequence_index is None:
            raise ValueError(
                "Não foi encontrada estabilização no intervalo avaliado."
            )

        optimal_index = stable_sequence_index + consecutive_points

        optimal_n = sample_sizes[optimal_index]
        optimal_se = standard_errors[optimal_index]
        optimal_allocation = allocations[optimal_index]

        summary = pd.DataFrame({
            "Categoria": strata,
            "Proporção na população": Wh,
            "Desvio-padrão": Sh,
            "Peso de Neyman": neyman_weights,
            "Quantidade na amostra": np.round(
                optimal_allocation
            ).astype(int)
        })

        summary["Percentual da amostra"] = (
            summary["Quantidade na amostra"]
            / summary["Quantidade na amostra"].sum()
        )

        strata_statistics = pd.DataFrame({
            "Categoria": strata,
            "Proporção na população": Wh,
            "Desvio-padrão": Sh,
            "Peso de Neyman": neyman_weights
        })

        if plot:
            plt.figure(figsize=(11, 6))

            plt.plot(
                sample_sizes,
                standard_errors,
                linewidth=3,
                label="Erro-padrão estratificado"
            )

            plt.scatter(
                optimal_n,
                optimal_se,
                s=120,
                label="Tamanho recomendado"
            )

            plt.axvline(
                optimal_n,
                linestyle="--"
            )

            plt.text(
                optimal_n,
                optimal_se,
                f"  n = {optimal_n}\n  EP = {optimal_se:.4f}",
                verticalalignment="bottom"
            )

            plt.xlabel("Tamanho total da amostra (n)")
            plt.ylabel("Erro-padrão estratificado (R$)")
            plt.title(
                "Estabilização marginal consecutiva "
                "do erro-padrão estratificado"
            )
            plt.grid(True)
            plt.legend()
            plt.show()

        results = {
            "optimal_n": int(optimal_n),
            "optimal_standard_error": float(optimal_se),
            "sample_sizes": sample_sizes,
            "standard_error_curve": standard_errors,
            "marginal_reduction": marginal_reduction,
            "stabilization_threshold": stabilization_threshold,
            "consecutive_points": consecutive_points,
            "strata_statistics": strata_statistics
        }

        return summary, results

    resumo_estratificado, resultado_estratificado = (
        optimal_stratified_sample_size(
            df=table18,
            value_column="TOTAL (R$)",
            strata_column="TIPO DA CONTA",
            n_min=100,
            n_max=100000,
            step=100,
            stabilization_threshold=0.001,
            consecutive_points=3,
            plot=True
        )
    )

    print(
        "Tamanho total recomendado:",
        resultado_estratificado["optimal_n"]
    )

    print(
        "Erro-padrão estratificado: R$ "
        f"{resultado_estratificado['optimal_standard_error']:.4f}"
    )

    display(
        resumo_estratificado.style.format({
            "Proporção na população": "{:.2%}",
            "Desvio-padrão": "{:.4f}",
            "Peso de Neyman": "{:.4f}",
            "Quantidade na amostra": "{:.0f}",
            "Percentual da amostra": "{:.2%}"
        })
    )


    quantidades = resumo_estratificado.set_index(
        "Categoria"
    )["Quantidade na amostra"]

    partes = []

    for i, categoria in enumerate(["ESSENTIAL", "PRIME", "VIP"]):

        grupo = table18[
            table18["TIPO DA CONTA"] == categoria
        ]

        parte = grupo.sample(
            n=int(quantidades[categoria]),
            replace=False,
            random_state=42 + i
        )

        partes.append(parte)

    amostra_estratificada = pd.concat(
        partes,
        ignore_index=True
    )

    print(
        "Total de registros selecionados:",
        len(amostra_estratificada)
    )

    print("\nRegistros por categoria:")
    print(
        amostra_estratificada["TIPO DA CONTA"].value_counts()
    )


        # Configurações do bootstrap.
    n_repeticoes = 5000
    nivel_confianca = 0.90
    rng = np.random.default_rng(42)


    pesos = resumo_estratificado.set_index(
        "Categoria"
    )["Proporção na população"]


    valores_por_categoria = {
        categoria: amostra_estratificada.loc[
            amostra_estratificada["TIPO DA CONTA"] == categoria,
            "TOTAL (R$)"
        ].to_numpy(dtype=float)
        for categoria in pesos.index
    }


    media_estimada = sum(
        pesos[categoria] * valores.mean()
        for categoria, valores in valores_por_categoria.items()
    )


    medias_bootstrap = np.zeros(n_repeticoes)

    for repeticao in range(n_repeticoes):

        media_repeticao = 0.0

        for categoria, valores in valores_por_categoria.items():


            reamostra = rng.choice(
                valores,
                size=len(valores),
                replace=True
            )


            media_repeticao += (
                pesos[categoria] * reamostra.mean()
            )

        medias_bootstrap[repeticao] = media_repeticao


    erro_padrao_bootstrap = medias_bootstrap.std(ddof=1)

    # Intervalo percentil: limites de 5% e 95% para confiança de 90%.
    alpha = 1 - nivel_confianca

    limite_inferior, limite_superior = np.quantile(
        medias_bootstrap,
        [alpha / 2, 1 - alpha / 2]
    )


    resultado_bootstrap_estratificado = {
        "tamanho_amostra": len(amostra_estratificada),
        "media_estimada": media_estimada,
        "erro_padrao": erro_padrao_bootstrap,
        "nivel_confianca": nivel_confianca,
        "limite_inferior": limite_inferior,
        "limite_superior": limite_superior,
        "n_repeticoes": n_repeticoes,
        "metodo_intervalo": "percentil"
    }


    plt.figure(figsize=(10, 6))

    sns.histplot(
        medias_bootstrap,
        bins=30,
        kde=True,
        stat="density"
    )

    plt.axvline(
        media_estimada,
        color="black",
        linestyle="-",
        label=f"Média estimada: R$ {media_estimada:.2f}"
    )

    plt.axvline(
        limite_inferior,
        color="red",
        linestyle="--",
        label=f"Limite inferior: R$ {limite_inferior:.2f}"
    )

    plt.axvline(
        limite_superior,
        color="red",
        linestyle="--",
        label=f"Limite superior: R$ {limite_superior:.2f}"
    )

    plt.title(
        "Bootstrap da média estratificada de TOTAL (R$)\n"
        "Amostra de 50.300 registros — confiança de 90%"
    )
    plt.xlabel("Média estratificada de TOTAL (R$)")
    plt.ylabel("Densidade")
    plt.legend()
    plt.tight_layout()
    plt.show()

Atividade1_1()
```

```python
def Atividade1_2():
    """
    Atividade A1.2 — Análise de outliers.

    OBJETIVO
    Identificar valores atípicos de TOTAL (R$) nas categorias
    ESSENTIAL, VIP e PRIME, considerando a distribuição de cada
    categoria separadamente.

    DADOS UTILIZADOS
    Base: table18, carregada previamente.
    Variável analisada: TOTAL (R$).
    Categorias identificadas pela coluna TIPO DA CONTA.

    MÉTODO
    Utiliza a regra de 1,5 vezes o intervalo interquartil (IQR).
    Para cada categoria, calcula:

        Q1: primeiro quartil, correspondente ao percentil 25.
        Q3: terceiro quartil, correspondente ao percentil 75.
        IQR = Q3 - Q1.
        Limite inferior = Q1 - 1,5 × IQR.
        Limite superior = Q3 + 1,5 × IQR.

    São classificados como outliers os valores estritamente
    inferiores ao limite inferior ou superiores ao limite superior.

    PROCEDIMENTOS
    1. Selecionar os registros de cada categoria.
    2. Converter TOTAL (R$) para valores numéricos e desconsiderar
       valores ausentes ou inválidos nesta análise.
    3. Calcular os quartis e os limites de identificação.
    4. Contar os outliers e calcular seu percentual em relação
       aos valores válidos da respectiva categoria.
    5. Exibir um boxplot e os resultados de cada categoria.

    RESULTADOS EXIBIDOS
    - Boxplots de ESSENTIAL, VIP e PRIME.
    - Quantidade de valores válidos por categoria.
    - Primeiro e terceiro quartis.
    - Intervalo interquartil.
    - Limites inferior e superior.
    - Quantidade e percentual de outliers.

    No total, foram identificados 31.410 registros outliers, equivalentes a
    aproximadamente 2,80% dos 1.120.000 registros analisados.
    """

    valores = pd.to_numeric(
        table18.loc[
            table18["TIPO DA CONTA"] == "ESSENTIAL",
            "TOTAL (R$)"
        ],
        errors="coerce"
    ).dropna()

    # Calcula os quartis e o intervalo interquartil.
    primeiro_quartil = valores.quantile(0.25)
    terceiro_quartil = valores.quantile(0.75)
    intervalo_interquartil = terceiro_quartil - primeiro_quartil

    # Define os limites para identificação dos outliers.
    limite_inferior = primeiro_quartil - 1.5 * intervalo_interquartil
    limite_superior = terceiro_quartil + 1.5 * intervalo_interquartil

    # Identifica os valores fora dos limites.
    outliers = valores[
        (valores < limite_inferior)
        | (valores > limite_superior)
    ]

    percentual_outliers = 100 * len(outliers) / len(valores)

    # Exibe a distribuição e os valores identificados como outliers.
    plt.figure(figsize=(12, 4))

    plt.boxplot(
        valores,
        vert=False,
        whis=1.5,
        patch_artist=True,
        boxprops={"facecolor": "#87CEEB"},
        medianprops={"color": "red", "linewidth": 2},
        flierprops={
            "marker": "o",
            "markersize": 3,
            "markerfacecolor": "#2878B5",
            "markeredgecolor": "#2878B5",
            "alpha": 0.3
        }
    )

    plt.title("Análise de outliers — ESSENTIAL")
    plt.xlabel("TOTAL (R$)")
    plt.yticks([1], ["ESSENTIAL"])
    plt.grid(axis="x", alpha=0.3)
    plt.tight_layout()
    plt.show()

    print("Categoria: ESSENTIAL")
    print("Quantidade de valores válidos:", len(valores))
    print(f"Primeiro quartil (Q1): R$ {primeiro_quartil:.2f}")
    print(f"Terceiro quartil (Q3): R$ {terceiro_quartil:.2f}")
    print(f"Intervalo interquartil: R$ {intervalo_interquartil:.2f}")
    print(f"Limite inferior: R$ {limite_inferior:.2f}")
    print(f"Limite superior: R$ {limite_superior:.2f}")
    print("Quantidade de outliers:", len(outliers))
    print(f"Percentual de outliers: {percentual_outliers:.2f}%")

    valores = pd.to_numeric(
        table18.loc[
            table18["TIPO DA CONTA"] == "VIP",
            "TOTAL (R$)"
        ],
        errors="coerce"
    ).dropna()


    primeiro_quartil = valores.quantile(0.25)
    terceiro_quartil = valores.quantile(0.75)
    intervalo_interquartil = terceiro_quartil - primeiro_quartil


    limite_inferior = primeiro_quartil - 1.5 * intervalo_interquartil
    limite_superior = terceiro_quartil + 1.5 * intervalo_interquartil


    outliers = valores[
        (valores < limite_inferior)
        | (valores > limite_superior)
    ]

    percentual_outliers = 100 * len(outliers) / len(valores)


    plt.figure(figsize=(12, 4))

    plt.boxplot(
        valores,
        vert=False,
        whis=1.5,
        patch_artist=True,
        boxprops={"facecolor": "#87CEEB"},
        medianprops={"color": "red", "linewidth": 2},
        flierprops={
            "marker": "o",
            "markersize": 3,
            "markerfacecolor": "#2878B5",
            "markeredgecolor": "#2878B5",
            "alpha": 0.3
        }
    )

    plt.title("Análise de outliers — VIP")
    plt.xlabel("TOTAL (R$)")
    plt.yticks([1], ["VIP"])
    plt.grid(axis="x", alpha=0.3)
    plt.tight_layout()
    plt.show()

    print("Categoria: VIP")
    print("Quantidade de valores válidos:", len(valores))
    print(f"Primeiro quartil (Q1): R$ {primeiro_quartil:.2f}")
    print(f"Terceiro quartil (Q3): R$ {terceiro_quartil:.2f}")
    print(f"Intervalo interquartil: R$ {intervalo_interquartil:.2f}")
    print(f"Limite inferior: R$ {limite_inferior:.2f}")
    print(f"Limite superior: R$ {limite_superior:.2f}")
    print("Quantidade de outliers:", len(outliers))
    print(f"Percentual de outliers: {percentual_outliers:.2f}%")

    valores = pd.to_numeric(
        table18.loc[
            table18["TIPO DA CONTA"] == "PRIME",
            "TOTAL (R$)"
        ],
        errors="coerce"
    ).dropna()


    primeiro_quartil = valores.quantile(0.25)
    terceiro_quartil = valores.quantile(0.75)
    intervalo_interquartil = terceiro_quartil - primeiro_quartil


    limite_inferior = primeiro_quartil - 1.5 * intervalo_interquartil
    limite_superior = terceiro_quartil + 1.5 * intervalo_interquartil


    outliers = valores[
        (valores < limite_inferior)
        | (valores > limite_superior)
    ]

    percentual_outliers = 100 * len(outliers) / len(valores)


    plt.figure(figsize=(12, 4))

    plt.boxplot(
        valores,
        vert=False,
        whis=1.5,
        patch_artist=True,
        boxprops={"facecolor": "#87CEEB"},
        medianprops={"color": "red", "linewidth": 2},
        flierprops={
            "marker": "o",
            "markersize": 3,
            "markerfacecolor": "#2878B5",
            "markeredgecolor": "#2878B5",
            "alpha": 0.3
        }
    )

    plt.title("Análise de outliers — PRIME")
    plt.xlabel("TOTAL (R$)")
    plt.yticks([1], ["PRIME"])
    plt.grid(axis="x", alpha=0.3)
    plt.tight_layout()
    plt.show()

    print("Categoria: PRIME")
    print("Quantidade de valores válidos:", len(valores))
    print(f"Primeiro quartil (Q1): R$ {primeiro_quartil:.2f}")
    print(f"Terceiro quartil (Q3): R$ {terceiro_quartil:.2f}")
    print(f"Intervalo interquartil: R$ {intervalo_interquartil:.2f}")
    print(f"Limite inferior: R$ {limite_inferior:.2f}")
    print(f"Limite superior: R$ {limite_superior:.2f}")
    print("Quantidade de outliers:", len(outliers))
    print(f"Percentual de outliers: {percentual_outliers:.2f}%")


Atividade1_2()
```

Nesta ETAPA, você deve fazer o **TESTE DE HIPÓTESE DE NORMALIDADE: Anderson-Darling' Test (AD test)** sobre:

```python
from scipy.stats import anderson


def Atividade1_3_1():
    """
    Atividade A1.3.1 — Teste de normalidade.

    OBJETIVO
    Avaliar a normalidade de TOTAL (R$) nas duas amostras,
    considerando ESSENTIAL, VIP e PRIME separadamente.

    DADOS UTILIZADOS
    amostra1 e amostra2, criadas previamente por amostragem
    estratificada com alocação de Neyman.

    MÉTODO
    Teste de Anderson-Darling para a distribuição normal,
    adotando nível de significância de 5%.

    HIPÓTESES
    H0: os valores seguem uma distribuição normal.
    H1: os valores não seguem uma distribuição normal.

    DECISÃO
    Estatística superior ao valor crítico: rejeitar H0.
    Caso contrário: não rejeitar H0.

    RESULTADOS EXIBIDOS
    Para cada categoria de cada amostra:
    quantidade de valores válidos, estatística do teste,
    valor crítico de 5% e conclusão.
    """

    for nome, amostra in [
        ("AMOSTRA 1", amostra1),
        ("AMOSTRA 2", amostra2)
    ]:
        print(f"\n{nome}")
        print("=" * 60)

        for categoria in ["ESSENTIAL", "VIP", "PRIME"]:
            valores = pd.to_numeric(
                amostra.loc[
                    amostra["TIPO DA CONTA"] == categoria,
                    "TOTAL (R$)"
                ],
                errors="coerce"
            ).dropna()

            resultado = anderson(valores, dist="norm")

            indice = np.flatnonzero(
                resultado.significance_level == 5
            )[0]

            valor_critico = resultado.critical_values[indice]

            print(f"\nCategoria: {categoria}")
            print("Quantidade de valores válidos:", len(valores))
            print(f"Estatística AD: {resultado.statistic:.4f}")
            print(f"Valor crítico a 5%: {valor_critico:.4f}")

            if resultado.statistic > valor_critico:
                print("Decisão: rejeitar H0.")
                print(
                    "Conclusão: há evidências de que TOTAL (R$) "
                    "não segue uma distribuição normal "
                    "nesta categoria."
                )
            else:
                print("Decisão: não rejeitar H0.")
                print(
                    "Conclusão: não há evidências suficientes "
                    "para rejeitar a normalidade de TOTAL (R$) "
                    "nesta categoria."
                )

Atividade1_3_1()
```

#### A1.3.2: Teste de Independência

Nesta ETAPA, você deve fazer o **TESTE DE ANOVA (+ Gráfico Bootstrap) sobre Testes de Independência: p_value** sobre:

*   SAMPLE 1 e SAMPLE 2
*   RESPONDER: Qual o BEST FIT ?

```python
def Atividade1_3_2():
    """
    Atividade A1.3.2 — Teste de independência.

    OBJETIVO
    Avaliar a independência entre TOTAL (R$) e TIPO DA CONTA.

    DADOS UTILIZADOS
    amostra1 e amostra2, criadas previamente por amostragem
    estratificada com alocação de Neyman.

    MÉTODO
    Teste de Qui-Quadrado
    adotando nível de significância de 5%.

    HIPÓTESES
    H0: as variáveis são independentes.
    H1: as variáveis não são independentes.

    DECISÃO
    Estatística superior ao valor crítico: rejeitar H0.
    Caso contrário: não rejeitar H0

    RESULTADOS EXIBIDOS
    Para cada categoria de cada amostra:
    quantidade de valores válidos, estatística do teste,
    valor crítico de 5% e conclusão.

    As amostras são independentes e O p-valor > 0.05 demonstra que o método de
    estratificação com alocação de Neyman garante a perfeita reprodução dos
    perfis Sem qualquer viés de seleção entre os grupos sorteados.
    """




    # Testamos se o tipo de conta é independente da Amostra (1 ou 2)

    contingencia = pd.crosstab(

        pd.concat([amostra1, amostra2], keys=['Amostra 1', 'Amostra 2']).index.get_level_values(0),

        pd.concat([amostra1, amostra2])['TIPO DA CONTA']

    )

    chi2, p_valor, dof, ex = chi2_contingency(contingencia)

    print("TESTE DE INDEPENDÊNCIA (Qui-Quadrado)")

    print("=" * 60)

    display(contingencia)

    print(f"Estatística Qui-Quadrado: {chi2:.4f}")

    print(f"valor do Teste: {p_valor:.6f}")

    if p_valor > 0.05:

        print("Decisão: Não rejeitar H0 (p > 0.05)")

        print("Conclusão: O perfil de clientes sorteado é independente do grupo amostral (Equivalência perfeita).")

    else:

        print("Decisão: Rejeitar H0 (p <= 0.05)")



    n_boot = 1000

    chi2_boot = []

    combinado = pd.concat([amostra1, amostra2])

    for _ in range(n_boot):

        boot_df = combinado.sample(frac=1.0, replace=True, random_state=None)

        boot_grupo = pd.crosstab(

            np.random.choice(['Amostra 1', 'Amostra 2'], size=len(boot_df)),

            boot_df['TIPO DA CONTA']

        )

        c_stat = chi2_contingency(boot_grupo)[0]

        chi2_boot.append(c_stat)

    plt.figure(figsize=(10, 5))

    sns.histplot(chi2_boot, kde=True, color='teal')

    plt.axvline(chi2, color='red', linestyle='--', label=f'Qui-Quadrado Observado: {chi2:.4f}')

    plt.title('Distribuição Bootstrap da Estatística Qui-Quadrado')

    plt.xlabel('Estatística Qui-Quadrado')

    plt.ylabel('Densidade')

    plt.legend()

    plt.grid(True, linestyle='--', alpha=0.5)

    plt.show()

Atividade1_3_2()
```

#### A1.3.3: Teste de Distribuição

Nesta ETAPA, você deve fazer o **TESTE DE ANOVA (+ Gráfico Bootstrap) sobre Teste de Mesma Distribuição: p_value** sobre:

*   SAMPLE 1 e SAMPLE 2
*   RESPONDER: Qual o BEST FIT ?

```python
from scipy.stats import ks_2samp

def Atividade1_3_3():
    """
    ATIVIDADE A1.3.3 — Teste de distribuição.

    OBJETIVO
    Avaliar a distribuição de TOTAL (R$) nas duas amostras,
    considerando ESSENTIAL, VIP e PRIME separadamente.

    DADOS UTILIZADOS
    amostra1 e amostra2, criadas previamente por amostragem
    estratificada com alocação de Neyman.

    MÉTODO
    Teste de Kolmogorov-Smirnov para a distribuição cumulativa,
    adotando nível de significância de 5%.

    HIPÓTESES
    H0: as distribuições são iguais (mesma distribuição cumulativa).
    H1: as distribuições são diferentes (distribuições cumulativas diferentes).

    Ambas as amostras seguem exatamente a mesma distribuição cumulativa, por tanto
    não há diferença significativa entre elas para definir BEST FIT.



    """


    valores1 = pd.to_numeric(amostra1['TOTAL (R$)'], errors='coerce').dropna().to_numpy()
    valores2 = pd.to_numeric(amostra2['TOTAL (R$)'], errors='coerce').dropna().to_numpy()

    # Teste KS para verificar se provêm da mesma distribuição cumulativa
    stat, p_value = ks_2samp(valores1, valores2)

    print("TESTE DE MESMA DISTRIBUIÇÃO (Kolmogorov-Smirnov)")
    print("=" * 60)
    print(f"Estatística KS: {stat:.4f}")
    print(f"p-valor do Teste: {p_value:.6f}")

    if p_value > 0.05:
        print("\nDecisão: Não rejeitar H0 (p > 0.05)")
        print("Conclusão: Ambas as amostras seguem exatamente a mesma distribuição cumulativa.")
    else:
        print("\nDecisão: Rejeitar H0")


    n_boot = 1000
    ks_stats = []

    for _ in range(n_boot):
        boot1 = np.random.choice(valores1, size=len(valores1), replace=True)
        boot2 = np.random.choice(valores2, size=len(valores2), replace=True)
        ks_stats.append(ks_2samp(boot1, boot2)[0])

    plt.figure(figsize=(10, 5))
    sns.histplot(ks_stats, kde=True, color='purple')
    plt.axvline(stat, color='red', linestyle='--', label=f'KS Observado: {stat:.4f}')
    plt.title('Distribuição Bootstrap da Estatística KS')
    plt.xlabel('Estatística KS')
    plt.ylabel('Densidade')
    plt.legend()
    plt.grid(True, linestyle='--', alpha=0.5)
    plt.show()

Atividade1_3_3()
```

#### A1.3.4: Teste das Médias Amostrais

Nesta ETAPA, você deve fazer o **TESTE DE ANOVA (+ Gráfico Bootstrap) sobre Teste das Médias Amostrais: p_value** sobre:

*   SAMPLE 1 e SAMPLE 2
*   RESPONDER: Qual o BEST FIT ?

```python
from scipy.stats import ttest_ind

def Atividade1_3_4():
    """
    Atividade1_3_4 — Teste de médias amostrais.

    OBJETIVO
    Avaliar a média de TOTAL (R$) nas duas amostras,
    considerando ESSENTIAL, VIP e PRIME separadamente.

    DADOS UTILIZADOS
    amostra1 e amostra2, criadas previamente por amostragem
    estratificada com alocação de Neyman.

    MÉTODO
    Teste de Médias Amostrais (Welch t-test)
    adotando nível de significância de 5%.

    HIPÓTESES
    H0: as médias amostrais possuem pouca diferença (média populacional equivalente).
    H1: as médias amostrais possuem diferença significativa (média populacional diferentes)

    As amostras possuem médias populacionais estatisticamente equivalentes, por
    tanto não há diferença significativa entre elas para definir BEST FIT.




    """

    valores1 = pd.to_numeric(amostra1['TOTAL (R$)'], errors='coerce').dropna().to_numpy()
    valores2 = pd.to_numeric(amostra2['TOTAL (R$)'], errors='coerce').dropna().to_numpy()

    # Teste de Médias (Welch t-test)
    stat, p_value = ttest_ind(valores1, valores2, equal_var=False)

    print("TESTE DE COMPARACÃO DE MÉDIAS AMOSTRAIS (Welch t-test)")
    print("=" * 60)
    print(f"Média Amostra 1: R$ {valores1.mean():.4f}")
    print(f"Média Amostra 2: R$ {valores2.mean():.4f}")
    print(f"Estatística t: {stat:.4f}")
    print(f"p-valor do Teste: {p_value:.6f}")

    if p_value > 0.05:
        print("Decisão: Não rejeitar H0 (p > 0.05)")
        print("Conclusão: Amostras possuem médias populacionais estatisticamente equivalentes.")
    else:
        print("Decisão: Rejeitar H0")

Atividade1_3_4()
```

* BEST FIT: O plano de amostragem de Neyman gerou grupos idênticos de representantes financeiros.

###Atividade A3.1: Discretizar a Variável Qualitativa (OPINIÃO DO CLIENTE)

```python
def Atividade3_1():
    def categorizar_opiniao(texto):
        if pd.isna(texto):
            return 'Neutro'
        texto = str(texto).lower()
        if any(w in texto for w in ['excelente', 'perfeito', 'otimo', 'ótimo', 'maravilhoso', 'superou', 'satisfeito', 'boa']):
            return 'Satisfeito'
        elif any(w in texto for w in ['ruim', 'pessimo', 'péssimo', 'demorou', 'insatisfeito', 'problema', 'defeito']):
            return 'Insatisfeito'
        return 'Neutro'

    amostra1['OPINIÃO_DISCRETA'] = amostra1['OPINIÃO DO CLIENTE'].apply(categorizar_opiniao)
    print("DISTRIBUIÇÃO DA OPINIÃO DISCRETIZADA DO CLIENTE")
    print("=" * 60)
    dist_op = amostra1['OPINIÃO_DISCRETA'].value_counts(normalize=True) * 100
    for categoria, pct in dist_op.items():
        print(f"{categoria:<15}: {pct:.2f}%")

Atividade3_1()
```

###Atividade A3.2: Análise Bootstrap do resultado da Pesquisa de Satisfação

```python
def Atividade3_2():
    n_boot = 1000
    valores = (amostra1['OPINIÃO_DISCRETA'] == 'Satisfeito').astype(int).to_numpy()
    proporcoes_boot = []

    for _ in range(n_boot):
        boot = np.random.choice(valores, size=len(valores), replace=True)
        proporcoes_boot.append(boot.mean())

    ic_inf, ic_sup = np.quantile(proporcoes_boot, [0.05, 0.95])

    plt.figure(figsize=(10, 5))
    sns.histplot(proporcoes_boot, kde=True, color='gold')
    plt.axvline(valores.mean(), color='black', label=f'Satisfação Média: {valores.mean()*100:.2f}%')
    plt.axvline(ic_inf, color='orange', linestyle='--', label=f'IC 90% Inf: {ic_inf*100:.2f}%')
    plt.axvline(ic_sup, color='orange', linestyle='--', label=f'IC 90% Sup: {ic_sup*100:.2f}%')
    plt.title('Bootstrap para Percentual de Clientes Satisfeitos (Confiança 90%)')
    plt.xlabel('Proporção de Clientes Satisfeitos')
    plt.ylabel('Frequência')
    plt.legend()
    plt.grid(True, linestyle='--', alpha=0.5)
    plt.show()

Atividade3_2()
```
